# AI Crop Doctor - Potato Disease Detection Model Training

This notebook trains a deep learning model for classifying Potato Leaf diseases using **Transfer Learning** with a pretrained **ResNet-18** network.

### Target Classes (3):
1. `Potato___Early_blight` (Fungal target-spot lesions)
2. `Potato___Late_blight` (Water-soaked dark lesions)
3. `Potato___healthy` (Healthy green leaves)

---

## 1. Setup & Environment Imports
Import PyTorch, TorchVision, evaluation metrics, and detect hardware acceleration (Apple Silicon MPS / CUDA GPU / CPU).

```python
import sys
import copy
from pathlib import Path
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, random_split
from torchvision import transforms, datasets, models
from sklearn.metrics import classification_report, confusion_matrix

# Add project root directory to path
ROOT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from config import settings
from modules.model import build_resnet_model, get_device

device = get_device()
print(f"Using compute device: '{device}'")
```

## 2. Dataset Loading & Train/Val/Test Splitting
Load 300 real leaf photos from `data/raw/plant_village_subset/` and split them into:
- **Train Set**: 70% (~210 images)
- **Validation Set**: 15% (~45 images)
- **Test Set**: 15% (~45 images)

```python
dataset_dir = settings.RAW_DATA_DIR / "plant_village_subset"
raw_dataset = datasets.ImageFolder(root=str(dataset_dir))
class_names = raw_dataset.classes

print(f"Detected Classes: {class_names}")
print(f"Total Images Loaded: {len(raw_dataset)}")

# Random split with fixed seed for reproducibility
generator = torch.Generator().manual_seed(42)
total_size = len(raw_dataset)
train_size = int(0.70 * total_size)
val_size = int(0.15 * total_size)
test_size = total_size - train_size - val_size

train_subset, val_subset, test_subset = random_split(
    raw_dataset, [train_size, val_size, test_size], generator=generator
)

print(f"Train size: {len(train_subset)} | Val size: {len(val_subset)} | Test size: {len(test_subset)}")
```

## 3. Data Transformations & Augmentation
Apply data augmentation (flips, rotations, color jitter) to training images to improve generalization, and standard preprocessing to validation/test sets.

```python
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=15),
    transforms.ColorJitter(brightness=0.1, contrast=0.1),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])

eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])

class TransformedDataset(torch.utils.data.Dataset):
    def __init__(self, subset, transform=None):
        self.subset = subset
        self.transform = transform
    def __getitem__(self, index):
        x, y = self.subset[index]
        if self.transform:
            x = self.transform(x)
        return x, y
    def __len__(self):
        return len(self.subset)

train_dataset = TransformedDataset(train_subset, transform=train_transform)
val_dataset = TransformedDataset(val_subset, transform=eval_transform)
test_dataset = TransformedDataset(test_subset, transform=eval_transform)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=16, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)
```

## 4. Transfer Learning Model Architecture (ResNet-18)
Load pretrained ResNet-18 backbone and update the final classification layer (`fc`) to output 3 class logits.

```python
model = build_resnet_model(num_classes=len(class_names), pretrained=True)
model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.0003)
print(model.fc)
```

## 5. Model Training & Validation Loop
Train model for 10 epochs while tracking train loss, train accuracy, val loss, and val accuracy. Save the best checkpoint.

```python
epochs = 10
best_model_wts = copy.deepcopy(model.state_dict())
best_val_loss = float("inf")

for epoch in range(1, epochs + 1):
    # Train
    model.train()
    r_loss, r_corrects = 0.0, 0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        _, preds = torch.max(outputs, 1)
        loss.backward()
        optimizer.step()
        r_loss += loss.item() * inputs.size(0)
        r_corrects += torch.sum(preds == labels.data).item()
    
    train_loss = r_loss / len(train_dataset)
    train_acc = (r_corrects / len(train_dataset)) * 100.0
    
    # Val
    model.eval()
    v_loss, v_corrects = 0.0, 0
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            _, preds = torch.max(outputs, 1)
            v_loss += loss.item() * inputs.size(0)
            v_corrects += torch.sum(preds == labels.data).item()
            
    val_loss = v_loss / len(val_dataset)
    val_acc = (v_corrects / len(val_dataset)) * 100.0
    
    print(f"Epoch {epoch:02d} | Train Loss: {train_loss:.4f} Acc: {train_acc:.2f}% | Val Loss: {val_loss:.4f} Acc: {val_acc:.2f}%")
    
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_model_wts = copy.deepcopy(model.state_dict())

# Save model checkpoint
save_dir = settings.BASE_DIR / "models" / "disease"
save_dir.mkdir(parents=True, exist_ok=True)
torch.save(best_model_wts, save_dir / "potato_disease_model.pth")
print(f"Saved best model to: {save_dir / 'potato_disease_model.pth'}")
```

## 6. Evaluation on Unseen Test Dataset
Load best checkpoint and evaluate accuracy, precision, recall, F1-score, and confusion matrix.

```python
model.load_state_dict(best_model_wts)
model.eval()

all_preds, all_targets = [], []
with torch.no_grad():
    for inputs, labels in test_loader:
        inputs = inputs.to(device)
        outputs = model(inputs)
        _, preds = torch.max(outputs, 1)
        all_preds.extend(preds.cpu().numpy())
        all_targets.extend(labels.numpy())

print("Classification Report:")
print(classification_report(all_targets, all_preds, target_names=class_names))
print("Confusion Matrix:")
print(confusion_matrix(all_targets, all_preds))
```
